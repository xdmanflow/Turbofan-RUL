# 03 — LSTM Sequence Model (30-cycle sliding windows)

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = '../data'
FIG_DIR = '../figures'

from src.train_lstm import run

model, history, pred, metrics = run(DATA_DIR, FIG_DIR, epochs=60)
metrics
```

## Training curves

```python
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='validation')
plt.xlabel('epoch'); plt.ylabel('MSE'); plt.legend(); plt.title('LSTM training')
plt.savefig(f'{FIG_DIR}/lstm_training.png', dpi=150, bbox_inches='tight')
```

## Predicted vs true RUL

```python
from src.data_prep import load_fd001

_, _, rul_true = load_fd001(DATA_DIR)
order = np.argsort(rul_true)
plt.figure(figsize=(12, 4))
plt.plot(rul_true[order], label='True RUL')
plt.plot(pred[order], label='LSTM prediction')
plt.xlabel('Test engines (sorted by true RUL)'); plt.ylabel('RUL'); plt.legend()
plt.savefig(f'{FIG_DIR}/lstm_pred_vs_true.png', dpi=150, bbox_inches='tight')
```

## Model comparison

```python
results = pd.concat([pd.read_csv(f'{FIG_DIR}/results_baselines.csv'), pd.read_csv(f'{FIG_DIR}/results_lstm.csv')])
results = results.sort_values('rmse').reset_index(drop=True)
display(results.round(2))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
results.plot.bar(x='model', y='rmse', ax=axes[0], legend=False, title='Test RMSE (lower is better)')
results.plot.bar(x='model', y='nasa_score', ax=axes[1], legend=False, title='NASA score (lower is better)')
plt.tight_layout(); plt.savefig(f'{FIG_DIR}/model_comparison.png', dpi=150, bbox_inches='tight')
```
