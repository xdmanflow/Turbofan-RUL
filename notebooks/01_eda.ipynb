# 01 — Exploratory Data Analysis (C-MAPSS FD001)

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = '../data'
FIG_DIR = '../figures'

from src.data_prep import SENSORS, SETTINGS, DROP_SENSORS, KEPT_SENSORS, add_rul, load_fd001

train, test, rul_true = load_fd001(DATA_DIR)
train = add_rul(train, cap=10_000)  # uncapped RUL for exploration
train.shape, test.shape
```

## Engine lifetimes

```python
lifetimes = train.groupby('unit')['cycle'].max()
print(lifetimes.describe().round(1))
lifetimes.plot.hist(bins=20, title='Engine lifetimes (cycles)')
plt.xlabel('cycles'); plt.savefig(f'{FIG_DIR}/lifetimes.png', dpi=150, bbox_inches='tight')
```

## Sensor variance — identifying near-constant sensors

```python
std = train[SETTINGS + SENSORS].std().sort_values()
print(std.round(5).to_string())
print('\nExcluded sensors:', DROP_SENSORS)
```

## Sensor trends over the life of a few engines

```python
fig, axes = plt.subplots(5, 3, figsize=(14, 14), sharex=True)
for ax, s in zip(axes.ravel(), KEPT_SENSORS):
    for u in train['unit'].unique()[:5]:
        g = train[train['unit'] == u]
        ax.plot(g['RUL'], g[s], lw=0.8)
    ax.set_title(s); ax.invert_xaxis()
fig.supxlabel('RUL (cycles)')
plt.tight_layout(); plt.savefig(f'{FIG_DIR}/sensor_trends.png', dpi=150, bbox_inches='tight')
```

## Correlation with RUL

```python
corr = train[KEPT_SENSORS + ['RUL']].corr()['RUL'].drop('RUL').sort_values()
corr.plot.barh(figsize=(6, 6), title='Correlation of each sensor with RUL')
plt.savefig(f'{FIG_DIR}/sensor_rul_correlation.png', dpi=150, bbox_inches='tight')
```
