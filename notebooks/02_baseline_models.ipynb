# 02 — Baseline Models (Linear Regression, Random Forest, Gradient Boosting)

```python
import sys
sys.path.append('..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = '../data'
FIG_DIR = '../figures'

from src.train_baselines import run

results, models, importances = run(DATA_DIR, FIG_DIR)
results.round(2)
```

## Feature importance (Random Forest)

```python
importances.plot.barh(figsize=(6, 6), title='Random Forest feature importance').invert_yaxis()
plt.savefig(f'{FIG_DIR}/rf_feature_importance.png', dpi=150, bbox_inches='tight')
```

## Predicted vs true RUL on the test set

```python
from src.data_prep import FEATURES, add_rul, fit_scaler, last_cycle, load_fd001, scale

train, test, rul_true = load_fd001(DATA_DIR)
scaler = fit_scaler(add_rul(train))
X_test = scale(last_cycle(test), scaler)[FEATURES]

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
for ax, (name, model) in zip(axes, models.items()):
    pred = model.predict(X_test).clip(min=0)
    ax.scatter(rul_true, pred, s=12)
    ax.plot([0, 150], [0, 150], 'k--', lw=1)
    ax.set_title(name); ax.set_xlabel('True RUL')
axes[0].set_ylabel('Predicted RUL')
plt.tight_layout(); plt.savefig(f'{FIG_DIR}/baselines_pred_vs_true.png', dpi=150, bbox_inches='tight')
```
